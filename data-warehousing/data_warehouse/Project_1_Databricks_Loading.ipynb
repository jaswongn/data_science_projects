```python
spark.sql("USE airline_project")
```

```python
%sql
-- Create Passenger Dimension Table
CREATE TABLE airline_project.Dim_Passenger (     
    Passenger_Key STRING,
    First_Name STRING,
	  Last_Name STRING,
	  Gender STRING,
	  Age INT,
	  Age_Type STRING,
    CONSTRAINT pk_dimpassenger PRIMARY KEY(Passenger_Key)
);
-- Create Nationality  Dimension Table
CREATE TABLE airline_project.Dim_Nationality (     
    Nationality_ID STRING,
    Nationality STRING,
    CONSTRAINT pk_dimnationality PRIMARY KEY(Nationality_ID)
);
-- Create Date Dimension Table
CREATE TABLE airline_project.Dim_Date (     
    Date_ID INT,
    Departure_Date Date,
	  Month INT,
	  Quarter INT,
	  YEAR INT,
    CONSTRAINT pk_dimdate PRIMARY KEY(Date_ID)
);
-- Create Location Dimension Table
CREATE TABLE airline_project.Dim_Location (     
    Location_ID INT,
    Region_Name STRING,
	  Country_Name STRING,
	  Continent STRING,
    CONSTRAINT pk_dimlocation PRIMARY KEY(Location_ID)
);
-- Create Airport Dimension Table
CREATE TABLE airline_project.Dim_Airport (     
    Airport_ID INT,
    Airport_Name STRING,
	  Airport_Size STRING,
    CONSTRAINT pk_dimairport PRIMARY KEY(Airport_ID)
);
-- Create Status Dimension Table
CREATE TABLE airline_project.Dim_Status (     
    Status_ID INT,
    Flight_Status STRING,
    CONSTRAINT pk_dimstatus PRIMARY KEY(Status_ID)
);

-- Create Flights Fact Table
CREATE TABLE airline_project.Fact_Flights (
    Flights_ID INT,
    Passenger_Key STRING,
    Nationality_ID STRING,
    Date_ID INT,
    Airport_ID INT,
    Location_ID INT,
    Status_ID INT,
    CONSTRAINT pk_dimfactflights PRIMARY KEY(Flights_ID),
    CONSTRAINT fk_dimpassenger FOREIGN KEY (Passenger_Key) REFERENCES Dim_Passenger(Passenger_Key),
	CONSTRAINT fk_dimnationality FOREIGN KEY (Nationality_ID) REFERENCES Dim_Nationality(Nationality_ID),
	CONSTRAINT fk_dimdate FOREIGN KEY (Date_ID) REFERENCES Dim_Date(Date_ID),
	CONSTRAINT fk_dimairport FOREIGN KEY (Airport_ID) REFERENCES Dim_Airport(Airport_ID),
	CONSTRAINT fk_dimlocation FOREIGN KEY (Location_ID) REFERENCES Dim_Location(Location_ID),
	CONSTRAINT fk_dimstatus FOREIGN KEY (Status_ID) REFERENCES Dim_Status(Status_ID)
);
```

```python
# read the .csv files and save as Spark DataFrames under dim_passenger_df, dim_nationality_df, dim_date_df, dim_location_df, dim_airport_df, dim_status_df, fact_flights_df
dim_passenger_df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .option("sep", ",") \
    .csv("/Volumes/workspace/airline_project/project_1/Dim_Passenger.csv");

dim_nationality_df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .option("sep", ",") \
    .csv("/Volumes/workspace/airline_project/project_1/Dim_Nationality.csv");

dim_date_df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .option("sep", ",") \
    .csv("/Volumes/workspace/airline_project/project_1/Dim_Date.csv");

dim_location_df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .option("sep", ",") \
    .csv("/Volumes/workspace/airline_project/project_1/Dim_Location.csv");

dim_airport_df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .option("sep", ",") \
    .csv("/Volumes/workspace/airline_project/project_1/Dim_Airport.csv");

dim_status_df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .option("sep", ",") \
    .csv("/Volumes/workspace/airline_project/project_1/Dim_Status.csv");

fact_flights_df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .option("sep", ",") \
    .csv("/Volumes/workspace/airline_project/project_1/Fact_Flights.csv");
```

```python
# populate the dim_passenger table with the data from the dim_passenger_df Spark DataFrame
dim_passenger_df.write \
    .mode("append") \
    .saveAsTable("airline_project.Dim_Passenger");
# populate the dim_nationality table with the data from the dim_nationality_df Spark DataFrame
dim_nationality_df.write \
    .mode("append") \
    .saveAsTable("airline_project.Dim_Nationality");
# populate the dim_date table with the data from the dim_date_df Spark DataFrame
dim_date_df.write \
    .mode("append") \
    .saveAsTable("airline_project.Dim_Date");
# populate the dim_location table with the data from the dim_location_df Spark DataFrame
dim_location_df.write \
    .mode("append") \
    .saveAsTable("airline_project.Dim_Location");
# populate the dim_airport table with the data from the dim_airport_df Spark DataFrame
dim_airport_df.write \
    .mode("append") \
    .saveAsTable("airline_project.Dim_Airport");
# populate the dim_status table with the data from the dim_status_df Spark DataFrame
dim_status_df.write \
    .mode("append") \
    .saveAsTable("airline_project.Dim_Status");
# populate the fact_flights table with the data from the fact_flights_df Spark DataFrame
fact_flights_df.write \
    .mode("append") \
    .saveAsTable("airline_project.Fact_Flights");
```
