```python
# Importing libraries
import pandas as pd
import numpy as np
import mlxtend
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
```

```python
# load in dataset
df = pd.read_csv('Final_Dataset.csv')
```

```python
# view dataset
df
```

```python
# dropping identifier columns
df = df.drop('Passenger_ID', axis = 'columns')
```

```python
# check if passenger id has been dropped
df.head()
```

```python
# create a function to add a Airport_Type attribute
def airport_type(airport):
    if airport in ['small_airport', 'medium_airport', 'large_airport']:
        return 'Commercial'
    else:
        return 'Non-Commercial'
# apply function to add airport type attribute
df['Airport_Type'] = df['Airport_Size'].apply(airport_type)
```

```python
# check if function was applied correctly
df.head(4)
```

```python
# add the header and then the class, this is for easier reading once we get the results
columns = ['Age_Type', 'Gender', 'Airport_Type', 'Continent' ,'Flight_Status']
new_df = df[['Age_Type', 'Gender', 'Airport_Type', 'Continent' ,'Flight_Status']]
for col in columns:
    new_df[col] = col +'=' + new_df[col].astype(str)
```

```python
new_df.head(4)
```

```python
#Data transformation
#TransactionEncoder() function only can handle string type
new_df = new_df.dropna()
new_df = new_df.astype(str)

#TransactionEncoder() was designed to covert lists to array
transactions = new_df.values.tolist()

#Covert the list to one-hot encoded boolean numpy array. 
#Apriori function allows boolean data type only, such as 1 and 0, or FALSE and TRUE.
te = TransactionEncoder()
array_te = te.fit_transform(transactions)

#Apriori function can handle dataframe only, covert the array to a dataframe
arm_df = pd.DataFrame(array_te, columns = te.columns_)
```

```python
arm_df.head(4)
```

```python
#Find the frequent itemsets
frequent_itemsets = apriori(arm_df,min_support=0.1,use_colnames =True)
# apply association rule mining based on lift = 1
rules = association_rules(frequent_itemsets, metric='lift', min_threshold=1.0)
# this one checks all RHS where it has flight_status
flight_status_rules = rules[rules['consequents'].apply(lambda x: all('Flight_Status=' in item for item in x))]
# output the top 3 rules of flight status based ordered by lift then confidence.
top_k_rules = flight_status_rules.sort_values(by=['lift','confidence'], ascending=False).head(3)
```

```python
# print out results
pd.set_option('display.max_colwidth', None)
print(top_k_rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```
