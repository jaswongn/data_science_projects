```python
# Triming strings, to make it more efficient, first create arracys for all the column names for each data set
airline_col = ['Passenger ID', 'First Name', 'Last Name', 'Gender', 'Nationality', 'Airport Name', 'Airport Country Code', 'Country Name', 'Airport Continent', 'Continents', 'Departure Date', 'Arrival Airport', 'Pilot Name', 'Flight Status']
airport_col = ['id', 'ident', 'type', 'name', 'latitude_deg', 'longitude_deg', 'elevation_ft', 'continent', 'iso_country', 'iso_region', 'municipality', 'scheduled_service', 'icao_code', 'iata_code', 'gps_code', 'local_code', 'home_link', 'wikipedia_link', 'keywords']
country_col = ['id', 'code', 'name', 'continent', 'wikipedia_link', 'keywords']
region_col = ['id', 'code', 'local_code', 'name', 'continent', 'iso_country', 'wikipedia_link', 'keywords']
```

```python
# trimming the airline dataset
Airline.drop_duplicates()
Airline_cleaned = Airline.copy()
for col in airline_col:
    Airline_cleaned[col] = Airline[col].astype(str).str.strip()

Airline_cleaned = Airline_cleaned.rename(columns={
    'Passenger ID': 'Passenger_ID',
    'First Name': 'First_Name',
    'Last Name': 'Last_Name',
    'Airport Name': 'Airport_Name_1',
    'Airport Country Code': 'Airport_Country_Code',
    'Country Name': 'Country_Name_1',
    'Airport Continent': 'Airport_Continent',
    'Continents': 'Continent_1',
    'Departure Date': 'Departure_Date',
    'Arrival Airport': 'Arrival_Airport',
    'Pilot Name': 'Pilot_Name',
    'Flight Status': 'Flight_Status'
})

Airline_cleaned.shape
```

```python
# trimming the airport dataset
Airport_cleaned = Airports.copy()
for col in airport_col:
    Airport_cleaned[col] = Airport_cleaned[col].astype(str).str.strip()

Airport_cleaned = Airport_cleaned.rename(columns={
    'id': 'Airport_ID',
    'type': 'Airport_Size',
    'name': 'Airport_Name_2',
    'latitude_deg': 'Latitude',
    'longitude_deg': 'Longitude',
    'continent': 'Continent_2',
    'iso_country': 'ISO_Country',
    'iso_region': 'ISO_Region',
    'iata_code': 'IATA_Code'
})
```

```python
# trimming the countries dataset
Countries_cleaned = Countries.copy()
for col in country_col:
    Countries_cleaned[col] = Countries_cleaned[col].astype(str).str.strip()

Countries_cleaned = Countries_cleaned.rename(columns={
    'id': 'Country_ID',
    'code': 'Country_Code',
    'name': 'Country_Name_3',
    'continent': 'Continent_3',
})
```

```python
# trimming the Regions dataset
Regions_cleaned = Regions.copy()
for col in region_col:
    Regions_cleaned[col] = Regions_cleaned[col].astype(str).str.strip()

Regions_cleaned = Regions_cleaned.rename(columns={
    'id': 'Regions_ID',
    'code': 'Regions_Code',
    'local_code': 'Local_Code',
    'name': 'Region_Name',
    'continent': 'Continent_4',
    'iso_country': 'ISO_Country_4'
})
```

```python
# code to drop duplicates and missing/na values
def clean_files(df, col):
    clean_file = df[df[col].notna() & df[col] != ""].drop_duplicates()
    return clean_file
```

```python
# we drop the missing values in the countries and regions dataset
Countries_cleaned_v2 = clean_files(Countries_cleaned, "Country_Code")
Regions_cleaned_v2 = clean_files(Regions_cleaned, "Regions_Code")
```

```python
# we then join the airport, countries and regions dataset with the airline dataset.
iata_match = Airline_cleaned.merge(Airport_cleaned, how='left', left_on='Arrival_Airport', right_on='IATA_Code')
matched = iata_match[iata_match['IATA_Code'].notna()]
unmatched = iata_match[iata_match['IATA_Code'].isna()].drop(columns=Airport_cleaned.columns)
valid_names = Airport_cleaned['Airport_Name_2'].value_counts()
valid_names = valid_names[valid_names == 1].index
filter_airports = Airport_cleaned[Airport_cleaned['Airport_Name_2'].isin(valid_names)]
name_matched = unmatched.merge(filter_airports, how='left', left_on='Airport_Name_1', right_on='Airport_Name_2')
Join_1 = pd.concat([matched, name_matched], ignore_index=True)
Join_2 = Join_1.merge(Regions_cleaned_v2, how='left', left_on='ISO_Region', right_on='Regions_Code')
Join_3 = Join_2.merge(Countries_cleaned_v2, how='left', left_on='ISO_Country_4', right_on='Country_Code')
```

```python
# we then select the main variables we will need for our dimensions and fact tables to answer our business query.
Joined_Dataset = Join_3[['Passenger_ID', 'Gender', 'First_Name', 'Last_Name', 'Age', 'Nationality', 'Airport_Name_2', 'Departure_Date', 'Flight_Status', 'Airport_Size', 'Continent_1', 'Country_Name_3', 'Region_Name']]
```

```python
# drop rows with missing rows
Joined_data_cleaned = Joined_Dataset.dropna(subset=['Airport_Name_2'])
```

```python
# finalizing the joined dataset and renaming variables before creating dimension tables
Final_Dataset = Joined_data_cleaned.rename(columns={
    'Airport_Name_2': 'Airport_Name',
    'Continent_1': 'Continent',
    'Country_Name_3': 'Country_Name'
})
Final_Dataset = Final_Dataset[['Passenger_ID', 'Gender', 'First_Name', 'Last_Name', 'Age', 'Nationality', 'Airport_Name', 'Departure_Date', 'Flight_Status', 'Airport_Size', 'Continent', 'Country_Name', 'Region_Name']]
```

```python
# check first 4 rows and see if headers were correctly renamed
Final_Dataset.head(4)
```

```python
# function used to create new variable, Age_Type, in passenger dimension table
def age_range(age):
    if age <= 17:
        return 'Child'
    elif (18 <= age) and (age <= 35):
        return 'Young_Adult'
    elif (36 <= age) and (age <= 59):
        return 'Adult'
    else:
        return 'Senior'
# create an Age_Type attribute in the dataset
Final_Dataset['Age_Type'] = Final_Dataset['Age'].apply(age_range)
# show the first 4 rows of the dataset
Final_Dataset.head(4)
```

```python
#Final_Dataset, exporting the final dataset
Final_Dataset.to_csv('Final_Dataset.csv', index = False)
```

```python
# passenger_df, removing duplicates
passenger_df = Final_Dataset[['Passenger_ID', 'First_Name', 'Last_Name', 'Gender', 'Age', 'Age_Type']].drop_duplicates()
# adding an id/key
p_index = range(101, 101+len(passenger_df))
passenger_df['Passenger_Key'] = ['A' + str(i) for i in p_index]
# rearrange the columns
passenger_df = passenger_df[['Passenger_Key','First_Name', 'Last_Name', 'Gender', 'Age', 'Age_Type']]
passenger_df.head()
```

```python
# nationality_df, removing duplicates
nationality_df = Final_Dataset[['Nationality']].drop_duplicates()
# adding an id/key
n_index = range(101, 101+len(nationality_df))
nationality_df['Nationality_ID'] = ['N' + str(i) for i in n_index]
# rearrange the columns
nationality_df = nationality_df[['Nationality_ID', 'Nationality']]
nationality_df.head()
```

```python
# creating a date dimension table
date_df = Final_Dataset[['Departure_Date']].drop_duplicates()
date_df['Month'] = pd.to_datetime(date_df['Departure_Date'], dayfirst = False).dt.month
date_df['Quarter'] = pd.to_datetime(date_df['Departure_Date'], dayfirst = False).dt.quarter
date_df['Year'] = pd.to_datetime(date_df['Departure_Date'], dayfirst = False).dt.year
date_df['Date_ID'] = range(1, len(date_df)+1)
date_df = date_df[['Date_ID', 'Departure_Date', 'Month', 'Quarter','Year']]
date_df.head(4)
```

```python
# location_df, removing duplicates
location_df = Final_Dataset[['Region_Name', 'Country_Name', 'Continent']].drop_duplicates()
# adding an id/key
location_df['Location_ID'] = range(1, len(location_df)+1)
# rearrange the columns
location_df = location_df[['Location_ID', 'Region_Name', 'Country_Name', 'Continent']]
location_df.head()
```

```python
# airport_df, removing duplicates
airport_df = Final_Dataset[['Airport_Name', 'Airport_Size']].drop_duplicates()
# adding an id/key
airport_df['Airport_ID'] = range(1, len(airport_df)+1)
# rearrange the columns
airport_df = airport_df[['Airport_ID', 'Airport_Name', 'Airport_Size']]
airport_df.head()
```

```python
# airport_df, removing duplicates
status_df = Final_Dataset[['Flight_Status']].drop_duplicates()
# adding an id/key
status_df['Status_ID'] = range(1, len(status_df)+1)
# rearrange the columns
status_df = status_df[['Status_ID', 'Flight_Status']]
status_df.head()
```

```python
#factless fact table
dfs = [passenger_df, nationality_df, date_df, airport_df, location_df, status_df]
flights_df = Final_Dataset.copy()
flights_df = flights_df.merge(passenger_df, on = ['First_Name','Last_Name', 'Age']).merge(nationality_df, on = ['Nationality']).merge(date_df, on=['Departure_Date']).merge(airport_df, on=['Airport_Name', 'Airport_Size']).merge(location_df, on=['Region_Name', 'Country_Name', 'Continent']).merge(status_df, on=['Flight_Status'])
flights_df
# adding a primary key to the fact table
flights_df['Flights_ID'] = range(1, len(flights_df)+1)
flights_df = flights_df[['Flights_ID', 'Passenger_Key', 'Nationality_ID', 'Date_ID', 'Airport_ID', 'Location_ID', 'Status_ID']]
flights_df.head()
```

```python
# converting to csv
csv_name = ['Dim_Passenger.csv','Dim_Nationality.csv', 'Dim_Date.csv', 'Dim_Airport.csv', 'Dim_Location.csv', 'Dim_Status.csv', 'Fact_Flights.csv']
dfs.append(flights_df)
for i in range(len(csv_name)):
    dfs[i].to_csv(csv_name[i], index = False, quoting=1, encoding='utf-8')
```
